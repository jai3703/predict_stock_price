```python
import requests
import pandas as pd

HEADERS = {
    "User-Agent": "Mozilla/5.0",
    "Accept-Language": "en-US,en;q=0.9",
}

session = requests.Session()
session.headers.update(HEADERS)

# Warm up session for NSE
session.get("https://www.nseindia.com", timeout=20)

shareholding_url = "https://www.nseindia.com/companies-listing/corporate-filings-shareholding-pattern"
pledge_url = "https://www.nseindia.com/companies-listing/corporate-filings-pledged-data"

shareholding_html = session.get(shareholding_url, timeout=30).text
pledge_html = session.get(pledge_url, timeout=30).text

# Try table extraction
shareholding_tables = pd.read_html(shareholding_html)
pledge_tables = pd.read_html(pledge_html)

print("Shareholding tables:", len(shareholding_tables))
print("Pledge tables:", len(pledge_tables))

# Usually inspect the largest/most relevant table
if shareholding_tables:
    sh_df = max(shareholding_tables, key=lambda x: len(x))
    print(sh_df.head())

if pledge_tables:
    pl_df = max(pledge_tables, key=lambda x: len(x))
    print(pl_df.head())
```

```python
def count_unique_substrings(s):
    # Dictionary to store the last seen index of each character
    last_seen = {}
    left = 0
    total_count = 0
    
    for right in range(len(s)):
        char = s[right]
        
        # If the character is in the current window, move 'left'
        # to one position after the previous occurrence of 'char'
        if char in last_seen and last_seen[char] >= left:
            left = last_seen[char] + 1
            
        # Update the last seen position of the character
        last_seen[char] = right
        
        # All substrings ending at 'right' and starting between 
        # 'left' and 'right' are valid.
        # Example: if window is "abc", valid endings are "c", "bc", "abc"
        total_count += (right - left + 1)
        
    return total_count

# Example from the image
s = "abac"
print(f"Total valid substrings for '{s}': {count_unique_substrings(s)}")
```
